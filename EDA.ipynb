## Set-up

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

path = "panel_data.csv"

merged_df = pd.read_csv(path)
```

## Explore missing data in each country

```python
# Create separate data frames for each country

US = merged_df[merged_df["Country"] == "United States"]
Japan = merged_df[merged_df["Country"] == "Japan"]
France = merged_df[merged_df["Country"] == "France"]
Italy = merged_df[merged_df["Country"] == "Italy"]
Canada = merged_df[merged_df["Country"] == "Canada"]
Germany = merged_df[merged_df["Country"] == "Germany"]
UK = merged_df[merged_df["Country"] == "United Kingdom"]
```

```python
# Create functions to summarise missing data

def missing_year(df,variable):
   data = df[df[variable].isna()]
   return data["Year"]

def missing_summary(df):
    missing_count = df.isna().sum()
    
    result = {}
    
    for variable in df.columns:
        if variable not in ["Country", "Year"]:
            if missing_count[variable] > 1:
                result[variable] = missing_year(df, variable)
                
    return(result)
```

## Visualisations

##### There is no Money Growth data for Italy, Germany, France and much recent data is missing for Canada. As a result the Money Growth variable will not be included in the dashboard visualising the relationship between the variables and inflation. Instead it will be included in the EDA for Japan, the United States and the United Kingdom.

```python
sns.set_style("whitegrid",{"grid.color": "grey", "grid.linestyle": "--", 
                           "grid.linewidth": 1.2})

sns.regplot(x="Money_growth", y="CPI_inflation", data=US, scatter_kws={"s": 100})

plt.xlabel("Money Growth")
plt.ylabel("Inflation")
plt.title("United States - Money Growth vs Inflation");
```

```python
sns.regplot(x="Money_growth", y="CPI_inflation", data=Japan, scatter_kws={"s": 100})

plt.xlabel("Money Growth")
plt.ylabel("Inflation")
plt.title("Japan - Money Growth vs Inflation");
```
